```python
df['Age']=df['Age'].fillna(df['Age'].median())
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['CabinAvailable']=df["Cabin"].notna().astype(int)
```

```python
df["Age"].hist(bins=30)
plt.title("Distribution of Passengers Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.countplot(data=df,x="Sex")
plt.title("Passanger Distribution by sex")
plt.show()
```

```python
sns.countplot(data=df,x="Survived")
plt.title("Survival Distribution")
plt.show()
```

```python
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Surivived by Gender")
plt.show()
```

```python
pd.crosstab(df["Sex"],df["Survived"],normalize="index")*100
```

```python
sns.countplot(data=df,x="Pclass",hue="Survived")
plt.title("Survival by Passanger Class")
plt.show()
```

```python
sns.boxplot(data=df,x="Survived",y="Age")
plt.title("Age Distribution by survival")
plt.show()
```

```python
sns.boxplot(data=df,x="Survived",y="Fare")
plt.title("Fare Distribution by survival")
plt.show()
```

```python
numeric_df=df.select_dtypes(include="number")
corr=numeric_df.corr()
plt.figure(figsize=(10,6))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df,
             vars=["Age","Fare","SibSp","Parch"],
             hue="Survived")
plt.show()
```
